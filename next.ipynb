```python
c1=input("enter first list of colors seperated by comma: ").split(',')
c2=input("enter second list of colors seperated by comma: ").split(',')
result=[]
for c in c1:
    if c not in c2:
        result.append(c)
print("first list:",c1)
print("second list:",c2)
print("colors in first and not in second list:",result)
```

```python
n = int(input("Enter number of items in dictionary: "))
my_dict = {}
for i in range(n):
    key = input(f"Enter key {i+1}: ")
    value = int(input(f"Enter value for {key}: "))
    my_dict[key] = value

asc = dict(sorted(my_dict.items()))
desc = dict(sorted(my_dict.items(), reverse=True))

print("Original Dictionary:", my_dict)
print("Ascending order:", asc)
print("Descending order:", desc)
```

```python
n1 = int(input("Enter number of items in dict1: "))
dict1 = {}
for i in range(n1):
    key = input(f"Enter key {i+1} for dict1: ")
    value = input(f"Enter value for {key}: ")
    dict1[key] = value

n2 = int(input("Enter number of items in dict2: "))
dict2 = {}
for i in range(n2):
    key = input(f"Enter key {i+1} for dict2: ")
    value = input(f"Enter value for {key}: ")
    dict2[key] = value

merged = {**dict1, **dict2}
print("Dict 1:", dict1)
print("Dict 2:", dict2)
print("Merged Dictionary:", merged)
```
