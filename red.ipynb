```python
import datetime
start=datetime.date.today().year
end=int(input("Enter end year"))
print(f"leap years between {start} and {end}")
for year in range(start,end+1):
    if(year%4==0 and year%100!=0) or (year%400==0):
        print(f"{year}  ")
```

```python
n = int(input("Enter number of terms: "))
a, b = 0, 1
print("Fibonacci series:")
for i in range(n):
    print(a, end=" ")
    a, b = b, a+b
print("\n")
```

```python
sq=lambda a:a*a
re=lambda l,b:l*b
tri=lambda l,h:0.5*l*h;


while True:
    print("\n 1.square \n 2.rectangle \n 3.triangle \n 4.exit \n")
    ch=int(input("enter your choice"))
    if ch==1:
        a=int(input("enter side"))
        print("area=",sq(a))
    elif ch==2:
        l=int(input("enter lenght"))
        b=int(input("enter breadth"))
        print("area= ",re(l,b))
    elif ch==3:
        l=int(input("enter base"))
        h=int(input("enter height"))
        print("area= ",tri(l,h))
    elif ch==4:
        break;
    else:
        print("invalid choivce")
        
```

```python
ls1 = list(map(int, input("Enter list of numbers separated by comma: ").split(',')))

ls2 = list(map(int, input("Enter list of numbers separated by comma: ").split(',')))
if len(ls1)==len(ls2):
    print("a) both list are of same length")
else:
    print("a) both are of different length")

#sum

if sum(ls1)==sum(ls2):
    print(("b) Sum are same"))
else:
    print(("b) Sum are not same"))

    
#intersect


common=set(ls1)&set(ls2)
if common:
    print("c) both have common values-- ",common)
else:
    print("c) both doesnt have common values")
```
